## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Research Question**

Which content pages should be reviewed first to maximize the effectiveness of the content refresh process?

**Decision**

The decision being improved is:

Which pages should an editor review first?

Who takes the action?

A content editor or SEO specialist reviews the ranked list and decides whether to:

Refresh outdated content,
Improve metadata,
Expand thin content,
Monitor performance,
Leave the page unchanged,


One content page (one row = one page).

**Output**

A ranked list of pages with

Refresh score,
Confidence score,
Reason codes explaining why the page was recommended.

**Cost of a Wrong Recommendation.**

A false positive wastes editor time reviewing pages that do not actually need attention.

A false negative means an important declining page may continue losing search visibility because it was never reviewed.

**Why ML helps**

Simple rules can capture obvious cases, but real content performance depends on many interacting signals such as visibility, age, CTR, engagement, search demand, and freshness.

Machine learning can combine these signals to prioritize pages more effectively than a fixed rule while still producing explainable recommendations.
